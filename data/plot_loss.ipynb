```python
import json
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from minicons import scorer


def plot_loss(file_path, start_step, end_step):
    # Load JSON data
    with open(file_path, 'r') as file:
        data = json.load(file)
    
    # Extract steps and loss values
    steps = []
    losses = []
    for entry in data['log_history']:
        if 'step' in entry and 'loss' in entry:
            step = entry['step']
            if start_step <= step <= end_step:
                steps.append(step)
                losses.append(entry['loss'])
    
    # Plot the data
    plt.figure(figsize=(24, 6))
    plt.plot(steps, losses, marker='o')
    plt.xlabel('Step')
    plt.ylabel('Loss')
    plt.title('Loss at Each Step')
    plt.grid(True)
    plt.xticks(np.arange(min(steps), max(steps)+1, 250))
    plt.show()

```

```python
# Example usage
plot_loss('trainer_state.json', start_step=2200, end_step=6000)
```
